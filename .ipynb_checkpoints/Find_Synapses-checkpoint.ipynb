```python
import tifffile as tiff
import matplotlib.pyplot as plt

filename = "c.tif"
movie = tiff.imread(filename)

fig, ax = plt.subplots()
frame_index = 0
im = ax.imshow(movie[frame_index], cmap='gray')
ax.set_title(f"Frame {frame_index}")

def on_key(event):
    global frame_index
    if event.key == 'right':
        frame_index = min(frame_index + 1, movie.shape[0] - 1)
    elif event.key == 'left':
        frame_index = max(frame_index - 1, 0)
    
    im.set_data(movie[frame_index])
    ax.set_title(f"Frame {frame_index}")
    fig.canvas.draw_idle()

fig.canvas.mpl_connect('key_press_event', on_key)
plt.show()
```

```python
import tifffile as tiff
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# -----------------------------
# Load movie
# -----------------------------
filename = "movie1.tif"
movie = tiff.imread(filename)   # expected shape: (T, Y, X)

print("Movie shape:", movie.shape)
print("Dtype:", movie.dtype)

T = movie.shape[0]

# -----------------------------
# Initial contrast limits
# -----------------------------
vmin_init = np.percentile(movie, 1)
vmax_init = np.percentile(movie, 99)

# -----------------------------
# Create widgets
# -----------------------------
frame_slider = widgets.IntSlider(
    value=0,
    min=0,
    max=T-1,
    step=1,
    description='Frame:',
    continuous_update=False
)

vmin_slider = widgets.FloatSlider(
    value=float(vmin_init),
    min=float(movie.min()),
    max=float(movie.max()),
    step=float((movie.max()-movie.min())/500),
    description='Min:'
)

vmax_slider = widgets.FloatSlider(
    value=float(vmax_init),
    min=float(movie.min()),
    max=float(movie.max()),
    step=float((movie.max()-movie.min())/500),
    description='Max:'
)

# -----------------------------
# Display function
# -----------------------------
def update(frame, vmin, vmax):
    plt.figure(figsize=(6,6))
    plt.imshow(movie[frame], cmap='gray', vmin=vmin, vmax=vmax)
    plt.title(f"Frame {frame}")
    plt.axis('off')
    plt.show()

widgets.interact(update,
                 frame=frame_slider,
                 vmin=vmin_slider,
                 vmax=vmax_slider);
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# Parameters
# -----------------------------
fs = 20.0
window_frames = int(0.5 * fs)   # 10 frames

# -----------------------------
# Load stim time series
# -----------------------------
df = pd.read_csv("stim1_mean_intensity_timeseries.csv")
stim = df["mean_intensity"].values

# -----------------------------
# Detect transitions
# -----------------------------
# Compute first difference
d_stim = np.diff(stim)

# Define threshold to detect real transitions
threshold = 0.1 * np.std(d_stim)

up_transitions = np.where(d_stim > threshold)[0]
down_transitions = np.where(d_stim < -threshold)[0]

print("Up transitions:", len(up_transitions))
print("Down transitions:", len(down_transitions))

# -----------------------------
# Collect post-transition frames
# -----------------------------
import tifffile as tiff

movie = tiff.imread("movie1_registered.tif")
T = movie.shape[0]

collected_frames = []

for idx in np.concatenate([up_transitions, down_transitions]):
    start = idx + 1
    end = start + window_frames
    
    if end < T:
        collected_frames.append(movie[start:end])

if len(collected_frames) == 0:
    raise ValueError("No valid transitions found.")

collected_frames = np.concatenate(collected_frames, axis=0)

print("Total collected frames:", collected_frames.shape[0])

# -----------------------------
# Compute average image
# -----------------------------
transition_avg_image = collected_frames.mean(axis=0)

# -----------------------------
# Display
# -----------------------------
vmin = np.percentile(transition_avg_image, 1)
vmax = np.percentile(transition_avg_image, 99)

plt.figure(figsize=(6,6))
plt.imshow(transition_avg_image, cmap='gray', vmin=vmin, vmax=vmax)
plt.title("Average Image: 0–0.5 s After Stim Transitions")
plt.axis('off')
plt.colorbar(label="Intensity")
plt.tight_layout()
plt.show()
```

```python
import tifffile as tiff
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# ============================================================
# Parameters
# ============================================================
filename = "movie1_registered.tif"
fs = 20.0  # Hz acquisition rate

# ============================================================
# Load movie
# ============================================================
movie = tiff.imread(filename)
T = movie.shape[0]
time = np.arange(T) / fs

print("Movie shape:", movie.shape)
print("Duration (s):", T/fs)

# ============================================================
# Compute global mean intensity trace
# ============================================================
frame_mean = movie.mean(axis=(1, 2))

# ============================================================
# Define exponential bleaching model
# F(t) = A * exp(-t/tau) + C
# ============================================================
def exp_decay(t, A, tau, C):
    return A * np.exp(-t / tau) + C

# Initial parameter guesses
A0 = frame_mean[0] - frame_mean[-1]
tau0 = T / fs / 2
C0 = frame_mean[-1]

p0 = [A0, tau0, C0]

# Fit curve
params, _ = curve_fit(exp_decay, time, frame_mean, p0=p0, maxfev=10000)
A, tau, C = params

fit_curve = exp_decay(time, A, tau, C)

print(f"Estimated bleaching tau: {tau:.2f} s")

# ============================================================
# Correct movie (multiplicative correction)
# ============================================================
fit_safe = np.maximum(fit_curve, 1e-8)  # prevent divide-by-zero
movie_corrected = movie / fit_safe[:, None, None]

# ============================================================
# Compute corrected mean trace
# ============================================================
corrected_mean = movie_corrected.mean(axis=(1, 2))

# ============================================================
# Plot results
# ============================================================
plt.figure(figsize=(10,4))

plt.plot(time, frame_mean, label="Raw mean")
plt.plot(time, fit_curve, label="Bleach fit", linewidth=2)
plt.plot(time, corrected_mean, label="Corrected mean")

plt.xlabel("Time (s)")
plt.ylabel("Mean Intensity (a.u.)")
plt.title("Bleach Correction")
plt.legend()
plt.tight_layout()
plt.show()

# ============================================================
# Save corrected movie
# ============================================================
tiff.imwrite(
    "movie1_registered_bleach_corrected.tif",
    movie_corrected.astype(np.float32)
)

print("Saved movie1_registered_bleach_corrected.tif")
```

```python
import numpy as np
import pandas as pd
import tifffile as tiff

# ============================================================
# Parameters
# ============================================================
fs = 20.0
baseline_windows = [(0,5), (55,60)]
transition_window_sec = 0.5
transition_frames = int(transition_window_sec * fs)

# ============================================================
# Load bleach-corrected movie
# ============================================================
movie = tiff.imread("movie1_registered_bleach_corrected.tif")
T, Y, X = movie.shape

print("Movie shape:", movie.shape)

# ============================================================
# Load stimulus time series
# ============================================================
df = pd.read_csv("stim1_mean_intensity_timeseries.csv")
stim = df["mean_intensity"].values

# ============================================================
# ---- BASELINE FRAMES ----
# ============================================================
baseline_indices = []

for start_s, end_s in baseline_windows:
    start_f = int(start_s * fs)
    end_f   = int(end_s * fs)
    
    if end_f > T:
        raise ValueError("Movie shorter than baseline window.")
    
    baseline_indices.extend(range(start_f, end_f))

baseline_indices = np.array(baseline_indices)

print("Baseline frames:", len(baseline_indices))

# ============================================================
# ---- TRANSITION FRAMES ----
# ============================================================
# Binary transition detection (robust)
stim_binary = stim > np.median(stim)
transitions = np.where(np.diff(stim_binary.astype(int)) != 0)[0]

transition_indices = []

for idx in transitions:
    start = idx + 1
    end = start + transition_frames
    
    if end < T:
        transition_indices.extend(range(start, end))

transition_indices = np.array(transition_indices)

print("Transition frames:", len(transition_indices))

# ============================================================
# ---- BUILD DISTRIBUTIONS ----
# ============================================================
# Shapes:
# baseline_dis:  (n_baseline_frames, Y, X)
# trans_dis:     (n_transition_frames, Y, X)

baseline_dis = movie[baseline_indices, :, :]
trans_dis    = movie[transition_indices, :, :]

print("Baseline distribution shape:", baseline_dis.shape)
print("Transition distribution shape:", trans_dis.shape)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------
# Select pixel
# ------------------------------------------------
y, x = 20, 20

baseline_pixel = baseline_dis[:, y, x]
trans_pixel    = trans_dis[:, y, x]

print("Baseline samples:", baseline_pixel.shape[0])
print("Transition samples:", trans_pixel.shape[0])

# ------------------------------------------------
# Plot histograms (density normalized)
# ------------------------------------------------
plt.figure(figsize=(8,5))

plt.hist(baseline_pixel, bins=40, density=True, alpha=0.5,
         label="Baseline")

plt.hist(trans_pixel, bins=40, density=True, alpha=0.5,
         label="Transition")

# Mean markers
plt.axvline(baseline_pixel.mean(), linestyle='--')
plt.axvline(trans_pixel.mean(), linestyle='--')

plt.xlabel("Pixel Intensity (Bleach Corrected)")
plt.ylabel("Probability Density")
plt.title("Pixel (20,20) Intensity Distributions")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------
# Means
# ------------------------------------------------
mean_base = baseline_dis.mean(axis=0)
mean_trans = trans_dis.mean(axis=0)

# ------------------------------------------------
# Variances (unbiased)
# ------------------------------------------------
var_base = baseline_dis.var(axis=0, ddof=1)
var_trans = trans_dis.var(axis=0, ddof=1)

# ------------------------------------------------
# Pooled standard deviation
# ------------------------------------------------
pooled_sd = np.sqrt(0.5 * (var_base + var_trans))

# Prevent divide-by-zero
pooled_sd = np.maximum(pooled_sd, 1e-12)

# ------------------------------------------------
# d' projection
# ------------------------------------------------
dprime_map = (mean_trans - mean_base) / pooled_sd

print("d' map shape:", dprime_map.shape)

plt.figure(figsize=(6,6))

vmax = np.percentile(np.abs(dprime_map), 99)

plt.imshow(dprime_map, cmap='bwr', vmin=-vmax, vmax=vmax)
plt.colorbar(label="d'")
plt.title("Pixel-wise d' Projection")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------
n_bins = 20  # histogram bins
eps = 1e-12  # numerical stability

Nb, Y, X = baseline_dis.shape
Nt = trans_dis.shape[0]

info_map = np.zeros((Y, X))

# ------------------------------------------------------------
# Compute information per pixel
# ------------------------------------------------------------
for y in range(Y):
    for x in range(X):
        
        base = baseline_dis[:, y, x]
        trans = trans_dis[:, y, x]
        
        # Combine samples
        values = np.concatenate([base, trans])
        labels = np.concatenate([
            np.zeros_like(base),
            np.ones_like(trans)
        ])
        
        # Histogram edges
        bins = np.linspace(values.min(), values.max(), n_bins+1)
        
        # Joint histogram
        joint = np.zeros((2, n_bins))
        
        for cls in [0, 1]:
            hist, _ = np.histogram(values[labels==cls], bins=bins)
            joint[cls] = hist
        
        joint = joint / joint.sum()
        
        # Marginals
        px = joint.sum(axis=0)
        py = joint.sum(axis=1)
        
        # Mutual information
        I = 0.0
        for i in range(2):
            for j in range(n_bins):
                if joint[i,j] > 0:
                    I += joint[i,j] * np.log2(
                        joint[i,j] / (py[i]*px[j] + eps)
                    )
        
        info_map[y,x] = I

print("Information map computed.")

plt.figure(figsize=(6,6))

vmax = np.percentile(info_map, 99)

plt.imshow(info_map, cmap='inferno', vmin=0, vmax=vmax)
plt.colorbar(label="Mutual Information (bits)")
plt.title("Pixel-wise Information Map")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import rankdata

Nb, Y, X = baseline_dis.shape
Nt = trans_dis.shape[0]

auc_map = np.zeros((Y, X))

# ------------------------------------------------------------
# Compute ROC AUC per pixel using rank formulation
# ------------------------------------------------------------
for y in range(Y):
    for x in range(X):
        
        base = baseline_dis[:, y, x]
        trans = trans_dis[:, y, x]
        
        combined = np.concatenate([base, trans])
        ranks = rankdata(combined)
        
        ranks_trans = ranks[Nb:]   # trans ranks
        
        # Mann–Whitney U statistic
        U = ranks_trans.sum() - Nt*(Nt+1)/2
        
        # AUC
        auc = U / (Nb * Nt)
        
        auc_map[y, x] = auc

print("ROC AUC map computed.")

plt.figure(figsize=(6,6))

# symmetric around 0.5 for visual clarity
vmax = np.percentile(np.abs(auc_map - 0.5), 99)
plt.imshow(auc_map, cmap='bwr', vmin=0.5 - vmax, vmax=0.5 + vmax)

plt.colorbar(label="ROC AUC")
plt.title("Pixel-wise ROC AUC Map")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import tifffile as tiff
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from scipy.stats import rankdata
from skimage.registration import phase_cross_correlation
from scipy.ndimage import shift

# ============================================================
# PARAMETERS
# ============================================================
fs = 20.0
baseline_windows = [(0,5), (55,60)]
transition_window_sec = 0.5
transition_frames = int(transition_window_sec * fs)

# ============================================================
# 1. LOAD MOVIES
# ============================================================
movie_raw = tiff.imread("movie1.tif")
stim_movie = tiff.imread("stim1.tif")

T, Y, X = movie_raw.shape
time = np.arange(T) / fs

print("Movie shape:", movie_raw.shape)

# ============================================================
# 2. RIGID REGISTRATION (phase cross-correlation)
# ============================================================
reference = movie_raw.mean(axis=0)
movie_reg = np.zeros_like(movie_raw)

for i in range(T):
    shift_est, _, _ = phase_cross_correlation(
        reference, movie_raw[i], upsample_factor=10
    )
    movie_reg[i] = shift(movie_raw[i], shift_est)

print("Registration complete.")

# ============================================================
# 3. BLEACH CORRECTION
# ============================================================
frame_mean = movie_reg.mean(axis=(1,2))

def exp_decay(t, A, tau, C):
    return A * np.exp(-t/tau) + C

p0 = [frame_mean[0]-frame_mean[-1], T/fs/2, frame_mean[-1]]
params, _ = curve_fit(exp_decay, time, frame_mean, p0=p0, maxfev=10000)
fit_curve = exp_decay(time, *params)

fit_safe = np.maximum(fit_curve, 1e-8)
movie_corr = movie_reg / fit_safe[:,None,None]

print("Bleach correction complete.")

# ============================================================
# 4. STIMULUS TRACE
# ============================================================
stim_trace = stim_movie.mean(axis=(1,2))
stim_binary = stim_trace > np.median(stim_trace)
transitions = np.where(np.diff(stim_binary.astype(int)) != 0)[0]

print("Transitions detected:", len(transitions))

# ============================================================
# 5. BASELINE INDICES
# ============================================================
baseline_idx = []

for start_s, end_s in baseline_windows:
    baseline_idx.extend(
        range(int(start_s*fs), int(end_s*fs))
    )

baseline_idx = np.array(baseline_idx)

# ============================================================
# 6. TRANSITION INDICES
# ============================================================
transition_idx = []

for idx in transitions:
    start = idx + 1
    end = start + transition_frames
    if end < T:
        transition_idx.extend(range(start, end))

transition_idx = np.array(transition_idx)

print("Baseline frames:", len(baseline_idx))
print("Transition frames:", len(transition_idx))

# ============================================================
# 7. BUILD DISTRIBUTIONS
# ============================================================
baseline_dis = movie_corr[baseline_idx,:,:]
trans_dis = movie_corr[transition_idx,:,:]

Nb = baseline_dis.shape[0]
Nt = trans_dis.shape[0]

# ============================================================
# 8. ROC AUC MAP
# ============================================================
auc_map = np.zeros((Y,X))

for y in range(Y):
    for x in range(X):
        base = baseline_dis[:,y,x]
        trans = trans_dis[:,y,x]
        
        combined = np.concatenate([base, trans])
        ranks = rankdata(combined)
        ranks_trans = ranks[Nb:]
        
        U = ranks_trans.sum() - Nt*(Nt+1)/2
        auc_map[y,x] = U / (Nb*Nt)

print("ROC AUC map computed.")

# ============================================================
# 9. DISPLAY RESULTS
# ============================================================

# ============================================================
# Plot bleach-corrected average time-series ONLY
# ============================================================

corrected_mean = movie_corr.mean(axis=(1,2))

plt.figure(figsize=(8,4))
plt.plot(time, corrected_mean)
plt.xlabel("Time (s)")
plt.ylabel("Mean Intensity (Bleach Corrected)")
plt.title("Bleach-Corrected Average Frame Intensity")
plt.tight_layout()
plt.show()

# ROC map
plt.figure(figsize=(6,6))
vmax = np.percentile(np.abs(auc_map - 0.5), 99)
plt.imshow(auc_map, cmap='bwr',
           vmin=0.5-vmax,
           vmax=0.5+vmax)
plt.colorbar(label="ROC AUC")
plt.title("Pixel-wise ROC AUC Map")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import tifffile as tiff
import matplotlib.pyplot as plt

from scipy.optimize import curve_fit
from scipy.ndimage import shift, gaussian_filter
from scipy.stats import rankdata
from skimage.registration import phase_cross_correlation
from skimage.feature import blob_log
from skimage.draw import disk

# ============================================================
# PARAMETERS
# ============================================================
fs = 20.0
baseline_windows = [(0,5), (55,60)]
transition_window_sec = 0.5
transition_frames = int(transition_window_sec * fs)

gaussian_sigma = 1.5  # matches ~2px half-width
min_blob_sigma = 1.0
max_blob_sigma = 2.5
blob_threshold_percentile = 99

# ============================================================
# 1. LOAD DATA
# ============================================================
movie_raw = tiff.imread("movie1.tif")
stim_movie = tiff.imread("stim1.tif")

T, Y, X = movie_raw.shape
time = np.arange(T) / fs

print("Movie shape:", movie_raw.shape)

# ============================================================
# 2. RIGID REGISTRATION
# ============================================================
reference = movie_raw.mean(axis=0)
movie_reg = np.zeros_like(movie_raw)

for i in range(T):
    shift_est, _, _ = phase_cross_correlation(
        reference, movie_raw[i], upsample_factor=10
    )
    movie_reg[i] = shift(movie_raw[i], shift_est)

print("Registration complete.")

# ============================================================
# 3. BLEACH CORRECTION (multiplicative)
# ============================================================
frame_mean = movie_reg.mean(axis=(1,2))

def exp_decay(t, A, tau, C):
    return A * np.exp(-t/tau) + C

p0 = [frame_mean[0]-frame_mean[-1], T/fs/2, frame_mean[-1]]
params, _ = curve_fit(exp_decay, time, frame_mean, p0=p0, maxfev=10000)
fit_curve = exp_decay(time, *params)

fit_safe = np.maximum(fit_curve, 1e-8)
movie_corr = movie_reg / fit_safe[:,None,None]

print("Bleach correction complete.")

# ============================================================
# 4. SPATIAL MATCHED FILTER
# ============================================================
movie_filt = np.zeros_like(movie_corr)

for t in range(T):
    movie_filt[t] = gaussian_filter(movie_corr[t], sigma=gaussian_sigma)

print("Spatial filtering complete.")

# ============================================================
# 5. STIMULUS TRANSITIONS
# ============================================================
stim_trace = stim_movie.mean(axis=(1,2))
stim_binary = stim_trace > np.median(stim_trace)
transitions = np.where(np.diff(stim_binary.astype(int)) != 0)[0]

# Baseline indices
baseline_idx = []
for start_s, end_s in baseline_windows:
    baseline_idx.extend(range(int(start_s*fs), int(end_s*fs)))
baseline_idx = np.array(baseline_idx)

# Transition indices
transition_idx = []
for idx in transitions:
    start = idx + 1
    end = start + transition_frames
    if end < T:
        transition_idx.extend(range(start, end))
transition_idx = np.array(transition_idx)

print("Baseline frames:", len(baseline_idx))
print("Transition frames:", len(transition_idx))

# ============================================================
# 6. RESPONSE MAP (ΔF)
# ============================================================
mean_base = movie_filt[baseline_idx].mean(axis=0)
mean_trans = movie_filt[transition_idx].mean(axis=0)

delta_map = mean_trans - mean_base

plt.figure(figsize=(6,6))
plt.imshow(delta_map, cmap='bwr')
plt.title("ΔF Response Map")
plt.axis('off')
plt.colorbar()
plt.show()

# ============================================================
# 7. BLOB DETECTION (LoG)
# ============================================================
threshold = np.percentile(delta_map, blob_threshold_percentile)

blobs = blob_log(
    delta_map,
    min_sigma=min_blob_sigma,
    max_sigma=max_blob_sigma,
    num_sigma=10,
    threshold=threshold
)

print("Blobs detected:", len(blobs))

# ============================================================
# 8. PER-BLOB ROC AUC
# ============================================================
synapses = []

for blob in blobs:
    y0, x0, sigma = blob
    
    r = int(2 * sigma)
    rr, cc = disk((int(y0), int(x0)), r, shape=delta_map.shape)
    
    base_vals = movie_filt[baseline_idx][:, rr, cc].mean(axis=1)
    trans_vals = movie_filt[transition_idx][:, rr, cc].mean(axis=1)
    
    Nb = len(base_vals)
    Nt = len(trans_vals)
    
    combined = np.concatenate([base_vals, trans_vals])
    ranks = rankdata(combined)
    ranks_trans = ranks[Nb:]
    
    U = ranks_trans.sum() - Nt*(Nt+1)/2
    auc = U / (Nb*Nt)
    
    synapses.append((y0, x0, auc))

print("Per-blob ROC computed.")

# ============================================================
# 9. DISPLAY DETECTED SYNAPSES
# ============================================================
plt.figure(figsize=(6,6))
plt.imshow(delta_map, cmap='gray')
plt.title("Detected Synapses (Red = AUC>0.7)")
plt.axis('off')

for y0, x0, auc in synapses:
    if auc > 0.7:
        plt.plot(x0, y0, 'ro', markersize=6)

plt.show()

print("Finished.")
```

```python
import numpy as np
import tifffile as tiff
import matplotlib.pyplot as plt

from scipy.optimize import curve_fit
from scipy.ndimage import shift, gaussian_filter
from scipy.stats import ks_2samp
from skimage.registration import phase_cross_correlation
from skimage.feature import peak_local_max
from skimage.draw import disk

# ============================================================
# PARAMETERS
# ============================================================
fs = 20.0
baseline_windows = [(0,5), (55,60)]
transition_window_sec = 0.5
transition_frames = int(transition_window_sec * fs)

sigma = 0.8
min_distance = 3
threshold_percentile = 98     # lowered for sensitivity
roi_radius = 2
alpha = 0.05                  # FDR level

# ============================================================
# 1. LOAD
# ============================================================
movie_raw = tiff.imread("movie1.tif")
stim_movie = tiff.imread("stim1.tif")

T, Y, X = movie_raw.shape
time = np.arange(T) / fs

# ============================================================
# 2. REGISTRATION
# ============================================================
reference = movie_raw.mean(axis=0)
movie_reg = np.zeros_like(movie_raw)

for i in range(T):
    shift_est, _, _ = phase_cross_correlation(
        reference, movie_raw[i], upsample_factor=10
    )
    movie_reg[i] = shift(movie_raw[i], shift_est)

# ============================================================
# 3. BLEACH CORRECTION
# ============================================================
frame_mean = movie_reg.mean(axis=(1,2))

def exp_decay(t, A, tau, C):
    return A * np.exp(-t/tau) + C

p0 = [frame_mean[0]-frame_mean[-1], T/fs/2, frame_mean[-1]]
params, _ = curve_fit(exp_decay, time, frame_mean, p0=p0, maxfev=10000)
fit_curve = exp_decay(time, *params)

movie_corr = movie_reg / np.maximum(fit_curve[:,None,None], 1e-8)

# ============================================================
# 4. STIMULUS TRANSITIONS
# ============================================================
stim_trace = stim_movie.mean(axis=(1,2))
stim_binary = stim_trace > np.median(stim_trace)
transitions = np.where(np.diff(stim_binary.astype(int)) != 0)[0]

baseline_idx = []
for start_s, end_s in baseline_windows:
    baseline_idx.extend(range(int(start_s*fs), int(end_s*fs)))
baseline_idx = np.array(baseline_idx)

transition_idx = []
for idx in transitions:
    start = idx + 1
    end = start + transition_frames
    if end < T:
        transition_idx.extend(range(start, end))
transition_idx = np.array(transition_idx)

# ============================================================
# 5. ΔF MAP (light smoothing)
# ============================================================
mean_base = movie_corr[baseline_idx].mean(axis=0)
mean_trans = movie_corr[transition_idx].mean(axis=0)
delta_map = gaussian_filter(mean_trans - mean_base, sigma=sigma)

# ============================================================
# 6. PEAK DETECTION
# ============================================================
coordinates = peak_local_max(
    delta_map,
    min_distance=min_distance,
    threshold_abs=np.percentile(delta_map, threshold_percentile)
)

print("Candidate peaks:", len(coordinates))

# ============================================================
# 7. KS TEST PER PEAK
# ============================================================
pvals = []
peaks = []

for y0, x0 in coordinates:
    
    rr, cc = disk((y0, x0), radius=roi_radius, shape=delta_map.shape)
    
    base_vals = movie_corr[baseline_idx][:, rr, cc].mean(axis=1)
    trans_vals = movie_corr[transition_idx][:, rr, cc].mean(axis=1)
    
    D, p = ks_2samp(base_vals, trans_vals)
    
    pvals.append(p)
    peaks.append((y0, x0, D, p))

pvals = np.array(pvals)

# ============================================================
# 8. FDR CORRECTION (Benjamini–Hochberg)
# ============================================================
if len(pvals) > 0:
    idx_sorted = np.argsort(pvals)
    sorted_p = pvals[idx_sorted]
    m = len(sorted_p)
    threshold_line = alpha * np.arange(1, m+1) / m
    
    significant = sorted_p <= threshold_line
    if np.any(significant):
        max_i = np.where(significant)[0].max()
        p_cutoff = sorted_p[max_i]
    else:
        p_cutoff = 0
    
    significant_peaks = [peaks[i] for i in range(len(peaks)) if pvals[i] <= p_cutoff]
else:
    significant_peaks = []

print("Significant synapses after FDR:", len(significant_peaks))

# ============================================================
# 9. DISPLAY NUMBERED SYNAPSES
# ============================================================
plt.figure(figsize=(8,8))
plt.imshow(delta_map, cmap='gray')
plt.title("KS-detected Synapses (FDR corrected)")
plt.axis('off')

for i, (y0, x0, D, p) in enumerate(significant_peaks, start=1):
    plt.plot(x0, y0, 'ro', markersize=6)
    plt.text(x0+1, y0+1, str(i), color='yellow', fontsize=9)

plt.show()
```

```python
import numpy as np
import tifffile as tiff
import matplotlib.pyplot as plt

from scipy.optimize import curve_fit
from scipy.ndimage import shift, gaussian_filter
from skimage.registration import phase_cross_correlation
from skimage.feature import peak_local_max

# ============================================================
# PARAMETERS
# ============================================================
fs = 20.0
baseline_windows = [(0,5), (55,60)]
transition_window_sec = 0.5
transition_frames = int(transition_window_sec * fs)

sigma_smooth = 0.8
min_distance = 3
threshold_percentile = 80      # LOWER peak threshold (permissive)
edge_margin = 5
patch_radius = 4               # for Gaussian fitting
roi_radius = 2                 # for ΔF/F extraction

min_sigma_fit = 0.6
max_sigma_fit = 3.0
min_r2 = 0.6                   # fit quality threshold

# ============================================================
# 1. LOAD DATA
# ============================================================
movie_raw = tiff.imread("movie1.tif")
stim_movie = tiff.imread("stim1.tif")

T, Y, X = movie_raw.shape
time = np.arange(T)/fs

print("Movie shape:", movie_raw.shape)

# ============================================================
# 2. RIGID REGISTRATION
# ============================================================
reference = movie_raw.mean(axis=0)
movie_reg = np.zeros_like(movie_raw)

for i in range(T):
    shift_est,_,_ = phase_cross_correlation(
        reference, movie_raw[i], upsample_factor=10
    )
    movie_reg[i] = shift(movie_raw[i], shift_est)

print("Registration complete.")

# ============================================================
# 3. BLEACH CORRECTION
# ============================================================
frame_mean = movie_reg.mean(axis=(1,2))

def exp_decay(t,A,tau,C):
    return A*np.exp(-t/tau)+C

p0 = [frame_mean[0]-frame_mean[-1], T/fs/2, frame_mean[-1]]
params,_ = curve_fit(exp_decay,time,frame_mean,p0=p0,maxfev=10000)
fit_curve = exp_decay(time,*params)

movie_corr = movie_reg / np.maximum(fit_curve[:,None,None],1e-8)

print("Bleach correction complete.")

# ============================================================
# 4. STIMULUS TRANSITIONS
# ============================================================
stim_trace = stim_movie.mean(axis=(1,2))
stim_binary = stim_trace > np.median(stim_trace)
transitions = np.where(np.diff(stim_binary.astype(int))!=0)[0]

baseline_idx=[]
for s,e in baseline_windows:
    baseline_idx.extend(range(int(s*fs),int(e*fs)))
baseline_idx=np.array(baseline_idx)

transition_idx=[]
for idx in transitions:
    start=idx+1
    end=start+transition_frames
    if end<T:
        transition_idx.extend(range(start,end))
transition_idx=np.array(transition_idx)

# ============================================================
# 5. ΔF MAP
# ============================================================
mean_base = movie_corr[baseline_idx].mean(axis=0)
mean_trans = movie_corr[transition_idx].mean(axis=0)
delta_map = gaussian_filter(mean_trans - mean_base, sigma=sigma_smooth)

# ============================================================
# 6. PERMISSIVE PEAK DETECTION
# ============================================================
threshold_abs = np.percentile(delta_map, threshold_percentile)

coords = peak_local_max(
    delta_map,
    min_distance=min_distance,
    threshold_abs=threshold_abs,
    exclude_border=False
)

print("Candidate peaks (before edge rejection):", len(coords))

# ============================================================
# 7. REJECT EDGE ARTEFACTS
# ============================================================
filtered_coords = []
for y0,x0 in coords:
    if (y0>edge_margin and y0<Y-edge_margin and
        x0>edge_margin and x0<X-edge_margin):
        filtered_coords.append((y0,x0))

print("After edge rejection:", len(filtered_coords))

# ============================================================
# 8. 2D GAUSSIAN MODEL
# ============================================================
def gaussian2d(coords,A,x0,y0,sigma,B):
    x,y = coords
    return A*np.exp(-((x-x0)**2+(y-y0)**2)/(2*sigma**2))+B

# ============================================================
# 9. FIT CANDIDATES
# ============================================================
synapses=[]

for y0,x0 in filtered_coords:
    
    y_min=max(0,y0-patch_radius)
    y_max=min(Y,y0+patch_radius+1)
    x_min=max(0,x0-patch_radius)
    x_max=min(X,x0+patch_radius+1)
    
    patch = delta_map[y_min:y_max, x_min:x_max]
    yy,xx = np.indices(patch.shape)
    
    A0 = patch.max()
    sigma0 = 1.5
    B0 = patch.min()
    
    try:
        popt,_ = curve_fit(
            gaussian2d,
            (xx,yy),
            patch.ravel(),
            p0=[A0,patch_radius,patch_radius,sigma0,B0]
        )
        
        A_fit,x_fit,y_fit,sigma_fit,B_fit = popt
        
        # Fit quality (R²)
        fitted = gaussian2d((xx,yy),*popt).reshape(patch.shape)
        ss_res = np.sum((patch-fitted)**2)
        ss_tot = np.sum((patch-np.mean(patch))**2)
        r2 = 1 - ss_res/ss_tot
        
        if (min_sigma_fit < sigma_fit < max_sigma_fit and
            r2 > min_r2 and
            A_fit > 0):
            
            # Extract ΔF/F
            rr,cc = np.indices((Y,X))
            mask = ((rr-y0)**2+(cc-x0)**2)<=roi_radius**2
            
            F0 = movie_corr[baseline_idx][:,mask].mean()
            F1 = movie_corr[transition_idx][:,mask].mean()
            
            dff = (F1-F0)/F0
            
            synapses.append((y0,x0,sigma_fit,A_fit,r2,dff))
            
    except:
        continue

print("Well-fit synapses:", len(synapses))

# ============================================================
# 10. DISPLAY NUMBERED SYNAPSES
# ============================================================
plt.figure(figsize=(8,8))
vmax = np.percentile(np.abs(delta_map),99)
plt.imshow(delta_map,cmap='bwr',vmin=-vmax,vmax=vmax)
plt.title("Detected Synapses (Gaussian Fit)")
plt.axis('off')

for i,(y0,x0,sigma_fit,A_fit,r2,dff) in enumerate(synapses,1):
    plt.plot(x0,y0,'o',markerfacecolor='none',
             markeredgecolor='white',markersize=10,markeredgewidth=2)
    plt.text(x0+2,y0+2,str(i),
             color='yellow',fontsize=12,fontweight='bold')

plt.show()

# ============================================================
# 11. PRINT ΔF/F PER SYNAPSE
# ============================================================
for i,(y0,x0,sigma_fit,A_fit,r2,dff) in enumerate(synapses,1):
    print(f"Synapse {i}: σ={sigma_fit:.2f}, R²={r2:.2f}, ΔF/F={dff:.3f}")
```

```python
import numpy as np
import tifffile as tiff
import matplotlib.pyplot as plt

from scipy.optimize import curve_fit
from scipy.ndimage import shift, gaussian_filter
from scipy.stats import ks_2samp
from skimage.registration import phase_cross_correlation
from skimage.feature import peak_local_max

# ============================================================
# PARAMETERS
# ============================================================
fs = 20.0
baseline_windows = [(0,5), (55,60)]

transition_frames = int(transition_window_sec * fs)

sigma_smooth = 0.8


edge_margin = 5


alpha = 0.05

threshold_percentile = 75      # much more permissive
min_distance = 2               # allow closer peaks
roi_radius = 3                 # better SNR
transition_window_sec = 0.75   # more samples for KS

# ============================================================
# 1. LOAD
# ============================================================
movie_raw = tiff.imread("movie1.tif")
stim_movie = tiff.imread("stim1.tif")

T, Y, X = movie_raw.shape
time = np.arange(T)/fs

# ============================================================
# 2. REGISTRATION
# ============================================================
reference = movie_raw.mean(axis=0)
movie_reg = np.zeros_like(movie_raw)

for i in range(T):
    shift_est,_,_ = phase_cross_correlation(
        reference, movie_raw[i], upsample_factor=10
    )
    movie_reg[i] = shift(movie_raw[i], shift_est)

# ============================================================
# 3. BLEACH CORRECTION
# ============================================================
frame_mean = movie_reg.mean(axis=(1,2))

def exp_decay(t,A,tau,C):
    return A*np.exp(-t/tau)+C

p0 = [frame_mean[0]-frame_mean[-1], T/fs/2, frame_mean[-1]]
params,_ = curve_fit(exp_decay,time,frame_mean,p0=p0,maxfev=10000)
fit_curve = exp_decay(time,*params)

movie_corr = movie_reg / np.maximum(fit_curve[:,None,None],1e-8)

# ============================================================
# 4. STIMULUS TRANSITIONS
# ============================================================
stim_trace = stim_movie.mean(axis=(1,2))
stim_binary = stim_trace > np.median(stim_trace)
transitions = np.where(np.diff(stim_binary.astype(int))!=0)[0]

baseline_idx=[]
for s,e in baseline_windows:
    baseline_idx.extend(range(int(s*fs),int(e*fs)))
baseline_idx=np.array(baseline_idx)

transition_idx=[]
for idx in transitions:
    start=idx+1
    end=start+transition_frames
    if end<T:
        transition_idx.extend(range(start,end))
transition_idx=np.array(transition_idx)

# ============================================================
# 5. ΔF MAP
# ============================================================
mean_base = movie_corr[baseline_idx].mean(axis=0)
mean_trans = movie_corr[transition_idx].mean(axis=0)
delta_map = gaussian_filter(mean_trans - mean_base, sigma=sigma_smooth)

# ============================================================
# 6. PEAK DETECTION
# ============================================================
threshold_abs = np.percentile(delta_map, threshold_percentile)

coords = peak_local_max(
    delta_map,
    min_distance=min_distance,
    threshold_abs=threshold_abs,
    exclude_border=False
)

# Reject edge peaks
filtered_coords=[]
for y0,x0 in coords:
    if (y0>edge_margin and y0<Y-edge_margin and
        x0>edge_margin and x0<X-edge_margin):
        filtered_coords.append((y0,x0))

print("Candidate peaks:",len(filtered_coords))

# ============================================================
# 7. KS TEST PER PEAK
# ============================================================
pvals=[]
peaks=[]

for y0,x0 in filtered_coords:
    
    rr,cc = np.indices((Y,X))
    mask = ((rr-y0)**2+(cc-x0)**2)<=roi_radius**2
    
    base_vals = movie_corr[baseline_idx][:,mask].mean(axis=1)
    trans_vals = movie_corr[transition_idx][:,mask].mean(axis=1)
    
    D,p = ks_2samp(base_vals,trans_vals)
    
    pvals.append(p)
    peaks.append((y0,x0,D,p))

pvals=np.array(pvals)

# ============================================================
# 8. FDR CORRECTION
# ============================================================
significant=[]

if len(pvals)>0:
    idx_sorted=np.argsort(pvals)
    sorted_p=pvals[idx_sorted]
    m=len(sorted_p)
    threshold_line=alpha*np.arange(1,m+1)/m
    significant_mask=sorted_p<=threshold_line
    
    if np.any(significant_mask):
        max_i=np.where(significant_mask)[0].max()
        p_cutoff=sorted_p[max_i]
        
        for i,(y0,x0,D,p) in enumerate(peaks):
            if p<=p_cutoff:
                significant.append((y0,x0))

print("Significant synapses:",len(significant))

# ============================================================
# 9. GAUSSIAN FIT (Morphology Only — Never Reject Activity)
# ============================================================

def gaussian2d(coords,A,x0,y0,sigma,B):
    x,y = coords
    return A*np.exp(-((x-x0)**2+(y-y0)**2)/(2*sigma**2))+B

synapses=[]

for y0,x0 in significant:
    
    sigma_fit = np.nan  # default if fit fails
    
    # Extract patch for morphology
    y_min=max(0,y0-patch_radius)
    y_max=min(Y,y0+patch_radius+1)
    x_min=max(0,x0-patch_radius)
    x_max=min(X,x0+patch_radius+1)
    
    patch = delta_map[y_min:y_max, x_min:x_max]
    yy,xx = np.indices(patch.shape)
    
    try:
        popt,_ = curve_fit(
            gaussian2d,
            (xx,yy),
            patch.ravel(),
            p0=[patch.max(), patch.shape[1]/2, patch.shape[0]/2, 1.5, patch.min()],
            maxfev=5000
        )
        sigma_fit = abs(popt[3])
    except:
        pass  # do NOT discard synapse
    
    # ΔF/F extraction (always done)
    rr,cc = np.indices((Y,X))
    mask = ((rr-y0)**2+(cc-x0)**2)<=roi_radius**2
    
    F0 = movie_corr[baseline_idx][:,mask].mean()
    F1 = movie_corr[transition_idx][:,mask].mean()
    dff = (F1-F0)/F0
    
    synapses.append((y0,x0,sigma_fit,dff))

# ============================================================
# 10. DISPLAY NUMBERED SYNAPSES
# ============================================================
plt.figure(figsize=(10,4))

vmin=np.percentile(delta_map,5)
vmax=np.percentile(delta_map,99)

plt.imshow(delta_map,cmap='gray',vmin=vmin,vmax=vmax)
plt.title(f"Detected Synapses (n={len(synapses)})")
plt.axis('off')

for i,(y0,x0,sigma_fit,dff) in enumerate(synapses,1):
    plt.scatter(x0,y0,s=60,c='red')
    plt.text(x0,y0,str(i),
             color='yellow',
             fontsize=9,
             ha='center',va='center',
             fontweight='bold')

plt.show()

# ============================================================
# 11. PRINT SUMMARY
# ============================================================
for i,(y0,x0,sigma_fit,dff) in enumerate(synapses,1):
    print(f"Synapse {i}: σ={sigma_fit:.2f}, ΔF/F={dff:.3f}")
```

```python
import numpy as np
import tifffile as tiff
import matplotlib.pyplot as plt

from scipy.optimize import curve_fit
from scipy.ndimage import shift, gaussian_filter
from scipy.stats import ks_2samp
from skimage.registration import phase_cross_correlation
from skimage.feature import peak_local_max

# ============================================================
# PARAMETERS (YOUR WORKING ONES)
# ============================================================
fs = 20.0

baseline_windows = [(0,5), (55,60)]
transition_window_sec = 0.75
threshold_percentile = 75
min_distance = 2
roi_radius = 3

sigma_smooth = 0.8
edge_margin = 5
patch_radius = 4
alpha = 0.05

baseline_window_dff = (1,5)

# ============================================================
# 1. LOAD
# ============================================================
movie_raw = tiff.imread("movie1.tif")
stim_movie = tiff.imread("stim1.tif")

T, Y, X = movie_raw.shape
time = np.arange(T)/fs

# ============================================================
# 2. REGISTRATION
# ============================================================
reference = movie_raw.mean(axis=0)
movie_reg = np.zeros_like(movie_raw)

for i in range(T):
    shift_est,_,_ = phase_cross_correlation(
        reference, movie_raw[i], upsample_factor=10
    )
    movie_reg[i] = shift(movie_raw[i], shift_est)

# ============================================================
# 3. BLEACH CORRECTION
# ============================================================
frame_mean = movie_reg.mean(axis=(1,2))

def exp_decay(t,A,tau,C):
    return A*np.exp(-t/tau)+C

p0 = [frame_mean[0]-frame_mean[-1], T/fs/2, frame_mean[-1]]
params,_ = curve_fit(exp_decay,time,frame_mean,p0=p0,maxfev=10000)
fit_curve = exp_decay(time,*params)

movie_corr = movie_reg / np.maximum(fit_curve[:,None,None],1e-8)

# ============================================================
# 4. STIM TRANSITIONS
# ============================================================
stim_trace = stim_movie.mean(axis=(1,2))
stim_binary = stim_trace > np.median(stim_trace)
transitions = np.where(np.diff(stim_binary.astype(int))!=0)[0]

baseline_idx=[]
for s,e in baseline_windows:
    baseline_idx.extend(range(int(s*fs),int(e*fs)))
baseline_idx=np.array(baseline_idx)

transition_frames = int(transition_window_sec*fs)
transition_idx=[]
for idx in transitions:
    start=idx+1
    end=start+transition_frames
    if end<T:
        transition_idx.extend(range(start,end))
transition_idx=np.array(transition_idx)

# ============================================================
# 5. ΔF MAP
# ============================================================
mean_base = movie_corr[baseline_idx].mean(axis=0)
mean_trans = movie_corr[transition_idx].mean(axis=0)
delta_map = gaussian_filter(mean_trans - mean_base, sigma=sigma_smooth)

# ============================================================
# 6. PEAK DETECTION
# ============================================================
threshold_abs = np.percentile(delta_map, threshold_percentile)

coords = peak_local_max(
    delta_map,
    min_distance=min_distance,
    threshold_abs=threshold_abs,
    exclude_border=False
)

filtered_coords=[]
for y0,x0 in coords:
    if (y0>edge_margin and y0<Y-edge_margin and
        x0>edge_margin and x0<X-edge_margin):
        filtered_coords.append((y0,x0))

print("Candidate peaks:",len(filtered_coords))

# ============================================================
# 7. KS + FDR DETECTION
# ============================================================
pvals=[]
peaks=[]

rr_full,cc_full = np.indices((Y,X))

for y0,x0 in filtered_coords:
    mask = ((rr_full-y0)**2+(cc_full-x0)**2)<=roi_radius**2
    
    base_vals = movie_corr[baseline_idx][:,mask].mean(axis=1)
    trans_vals = movie_corr[transition_idx][:,mask].mean(axis=1)
    
    D,p = ks_2samp(base_vals,trans_vals)
    
    pvals.append(p)
    peaks.append((y0,x0,D,p))

pvals=np.array(pvals)

significant=[]
if len(pvals)>0:
    idx_sorted=np.argsort(pvals)
    sorted_p=pvals[idx_sorted]
    m=len(sorted_p)
    threshold_line=alpha*np.arange(1,m+1)/m
    significant_mask=sorted_p<=threshold_line
    
    if np.any(significant_mask):
        max_i=np.where(significant_mask)[0].max()
        p_cutoff=sorted_p[max_i]
        for i,(y0,x0,D,p) in enumerate(peaks):
            if p<=p_cutoff:
                significant.append((y0,x0))

print("Significant synapses:",len(significant))

# ============================================================
# 8. GAUSSIAN FIT (MORPHOLOGY ONLY)
# ============================================================
def gaussian2d(coords,A,x0,y0,sigma,B):
    x,y = coords
    return A*np.exp(-((x-x0)**2+(y-y0)**2)/(2*sigma**2))+B

synapses=[]

for y0,x0 in significant:
    
    sigma_fit = 1.5
    
    y_min=max(0,y0-patch_radius)
    y_max=min(Y,y0+patch_radius+1)
    x_min=max(0,x0-patch_radius)
    x_max=min(X,x0+patch_radius+1)
    
    patch = delta_map[y_min:y_max, x_min:x_max]
    yy,xx = np.indices(patch.shape)
    
    try:
        popt,_ = curve_fit(
            gaussian2d,
            (xx,yy),
            patch.ravel(),
            p0=[patch.max(), patch.shape[1]/2, patch.shape[0]/2, 1.5, patch.min()],
            maxfev=5000
        )
        sigma_fit = abs(popt[3])
    except:
        pass
    
    # Detection ΔF/F stored
    rr,cc = np.indices((Y,X))
    mask = ((rr-y0)**2+(cc-x0)**2)<=roi_radius**2
    F0 = movie_corr[baseline_idx][:,mask].mean()
    F1 = movie_corr[transition_idx][:,mask].mean()
    dff_detect = (F1-F0)/F0
    
    synapses.append((y0,x0,sigma_fit,dff_detect))

# ============================================================
# 9. SELECT TOP 5 (YOUR WORKING METHOD)
# ============================================================
synapses_sorted = sorted(synapses, key=lambda x: abs(x[3]), reverse=True)
top_synapses = synapses_sorted[:5]

print("Top 5 synapses:")
for i,(y,x,sigma,dff) in enumerate(top_synapses,1):
    print(f"{i}: (y={y}, x={x}), sigma={sigma:.2f}")

# ============================================================
# 10. COMPUTE ΔF/F TRACES (YOUR GOOD VERSION)
# ============================================================
dff_traces = []

for y0,x0,sigma_fit,_ in top_synapses:
    
    if np.isnan(sigma_fit):
        sigma_fit = 1.5
    
    patch_radius = int(np.ceil(3*sigma_fit))
    
    y_min=max(0,y0-patch_radius)
    y_max=min(Y,y0+patch_radius+1)
    x_min=max(0,x0-patch_radius)
    x_max=min(X,x0+patch_radius+1)
    
    yy,xx = np.indices((y_max-y_min, x_max-x_min))
    
    G = np.exp(-((xx-(x0-x_min))**2 + (yy-(y0-y_min))**2) /
               (2*sigma_fit**2))
    G = G.ravel()
    
    amplitude_trace = []
    
    for t in range(T):
        patch = movie_corr[t,y_min:y_max,x_min:x_max].ravel()
        A_matrix = np.vstack([G, np.ones_like(G)]).T
        coef,_,_,_ = np.linalg.lstsq(A_matrix, patch, rcond=None)
        amplitude_trace.append(coef[0])
    
    amplitude_trace = np.array(amplitude_trace)
    
    baseline_idx_dff = np.where(
        (time >= baseline_window_dff[0]) &
        (time <= baseline_window_dff[1])
    )[0]
    
    F0 = amplitude_trace[baseline_idx_dff].mean()
    dff = (amplitude_trace - F0) / F0
    
    dff_traces.append(dff)

# ============================================================
# 11. PLOT TOP 5
# ============================================================
plt.figure(figsize=(10,6))

for i,trace in enumerate(dff_traces):
    plt.plot(time, trace, label=f"Synapse {i+1}")

plt.xlabel("Time (s)")
plt.ylabel("ΔF/F (baseline 1–5 s)")
plt.title("Gaussian-Fit Synapse Responses (Top 5)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import tifffile as tiff
import matplotlib.pyplot as plt

from scipy.optimize import curve_fit
from scipy.ndimage import shift, gaussian_filter
from scipy.stats import ks_2samp
from skimage.registration import phase_cross_correlation
from skimage.feature import peak_local_max

# ============================================================
# PARAMETERS (YOUR WORKING ONES)
# ============================================================
fs = 20.0

baseline_windows = [(0,5), (55,60)]
transition_window_sec = 0.75
threshold_percentile = 75
min_distance = 2
roi_radius = 3

sigma_smooth = 0.8
edge_margin = 5
patch_radius = 4
alpha = 0.05

baseline_window_dff = (1,5)

# ============================================================
# 1. LOAD
# ============================================================
movie_raw = tiff.imread("movie1.tif")
stim_movie = tiff.imread("stim1.tif")

T, Y, X = movie_raw.shape
time = np.arange(T)/fs

# ============================================================
# 2. REGISTRATION
# ============================================================
reference = movie_raw.mean(axis=0)
movie_reg = np.zeros_like(movie_raw)

for i in range(T):
    shift_est,_,_ = phase_cross_correlation(
        reference, movie_raw[i], upsample_factor=10
    )
    movie_reg[i] = shift(movie_raw[i], shift_est)

# ============================================================
# 3. BLEACH CORRECTION
# ============================================================
frame_mean = movie_reg.mean(axis=(1,2))

def exp_decay(t,A,tau,C):
    return A*np.exp(-t/tau)+C

p0 = [frame_mean[0]-frame_mean[-1], T/fs/2, frame_mean[-1]]
params,_ = curve_fit(exp_decay,time,frame_mean,p0=p0,maxfev=10000)
fit_curve = exp_decay(time,*params)

movie_corr = movie_reg / np.maximum(fit_curve[:,None,None],1e-8)

# ============================================================
# 4. STIM TRANSITIONS
# ============================================================
stim_trace = stim_movie.mean(axis=(1,2))
stim_binary = stim_trace > np.median(stim_trace)
transitions = np.where(np.diff(stim_binary.astype(int))!=0)[0]

baseline_idx=[]
for s,e in baseline_windows:
    baseline_idx.extend(range(int(s*fs),int(e*fs)))
baseline_idx=np.array(baseline_idx)

transition_frames = int(transition_window_sec*fs)
transition_idx=[]
for idx in transitions:
    start=idx+1
    end=start+transition_frames
    if end<T:
        transition_idx.extend(range(start,end))
transition_idx=np.array(transition_idx)

# ============================================================
# 5. ΔF MAP
# ============================================================
mean_base = movie_corr[baseline_idx].mean(axis=0)
mean_trans = movie_corr[transition_idx].mean(axis=0)
delta_map = gaussian_filter(mean_trans - mean_base, sigma=sigma_smooth)

# ============================================================
# 6. PEAK DETECTION
# ============================================================
threshold_abs = np.percentile(delta_map, threshold_percentile)

coords = peak_local_max(
    delta_map,
    min_distance=min_distance,
    threshold_abs=threshold_abs,
    exclude_border=False
)

filtered_coords=[]
for y0,x0 in coords:
    if (y0>edge_margin and y0<Y-edge_margin and
        x0>edge_margin and x0<X-edge_margin):
        filtered_coords.append((y0,x0))

print("Candidate peaks:",len(filtered_coords))

# ============================================================
# 7. KS + FDR DETECTION
# ============================================================
pvals=[]
peaks=[]
rr_full,cc_full = np.indices((Y,X))

for y0,x0 in filtered_coords:
    mask = ((rr_full-y0)**2+(cc_full-x0)**2)<=roi_radius**2
    
    base_vals = movie_corr[baseline_idx][:,mask].mean(axis=1)
    trans_vals = movie_corr[transition_idx][:,mask].mean(axis=1)
    
    D,p = ks_2samp(base_vals,trans_vals)
    
    pvals.append(p)
    peaks.append((y0,x0,D,p))

pvals=np.array(pvals)

significant=[]
if len(pvals)>0:
    idx_sorted=np.argsort(pvals)
    sorted_p=pvals[idx_sorted]
    m=len(sorted_p)
    threshold_line=alpha*np.arange(1,m+1)/m
    significant_mask=sorted_p<=threshold_line
    
    if np.any(significant_mask):
        max_i=np.where(significant_mask)[0].max()
        p_cutoff=sorted_p[max_i]
        for i,(y0,x0,D,p) in enumerate(peaks):
            if p<=p_cutoff:
                significant.append((y0,x0))

print("Significant synapses:",len(significant))

# ============================================================
# 8. GAUSSIAN FIT (MORPHOLOGY ONLY)
# ============================================================
def gaussian2d(coords,A,x0,y0,sigma,B):
    x,y = coords
    return A*np.exp(-((x-x0)**2+(y-y0)**2)/(2*sigma**2))+B

synapses=[]

for y0,x0 in significant:
    
    sigma_fit = 1.5
    
    y_min=max(0,y0-patch_radius)
    y_max=min(Y,y0+patch_radius+1)
    x_min=max(0,x0-patch_radius)
    x_max=min(X,x0+patch_radius+1)
    
    patch = delta_map[y_min:y_max, x_min:x_max]
    yy,xx = np.indices(patch.shape)
    
    try:
        popt,_ = curve_fit(
            gaussian2d,
            (xx,yy),
            patch.ravel(),
            p0=[patch.max(), patch.shape[1]/2, patch.shape[0]/2, 1.5, patch.min()],
            maxfev=5000
        )
        sigma_fit = abs(popt[3])
    except:
        pass
    
    rr,cc = np.indices((Y,X))
    mask = ((rr-y0)**2+(cc-x0)**2)<=roi_radius**2
    F0 = movie_corr[baseline_idx][:,mask].mean()
    F1 = movie_corr[transition_idx][:,mask].mean()
    dff_detect = (F1-F0)/F0
    
    synapses.append((y0,x0,sigma_fit,dff_detect))

# ============================================================
# 9. PLOT SYNAPSE LOCATIONS
# ============================================================
plt.figure(figsize=(8,4))

vmin=np.percentile(delta_map,5)
vmax=np.percentile(delta_map,99)

plt.imshow(delta_map,cmap='gray',vmin=vmin,vmax=vmax)
plt.title(f"Detected Synapses (n={len(synapses)})")
plt.axis('off')

for i,(y0,x0,sigma_fit,dff) in enumerate(synapses,1):
    plt.scatter(x0,y0,s=60,c='red')
    plt.text(x0,y0,str(i),
             color='yellow',
             fontsize=9,
             ha='center',va='center',
             fontweight='bold')

plt.tight_layout()
plt.show()

# ============================================================
# 10. SELECT TOP 5
# ============================================================
synapses_sorted = sorted(synapses, key=lambda x: abs(x[3]), reverse=True)
top_synapses = synapses_sorted[:5]

# ============================================================
# 11. COMPUTE ΔF/F TRACES  (Optimised, Stable Version)
# ============================================================

dff_traces = []

for y0,x0,sigma_fit,_ in top_synapses:
    
    if np.isnan(sigma_fit):
        sigma_fit = 1.5
    
    patch_radius = int(np.ceil(3*sigma_fit))
    
    y_min = max(0,y0-patch_radius)
    y_max = min(Y,y0+patch_radius+1)
    x_min = max(0,x0-patch_radius)
    x_max = min(X,x0+patch_radius+1)
    
    # ---- Precompute Gaussian template ONCE ----
    yy,xx = np.indices((y_max-y_min, x_max-x_min))
    
    G = np.exp(
        -((xx-(x0-x_min))**2 + (yy-(y0-y_min))**2) /
        (2*sigma_fit**2)
    )
    
    G = G.ravel()
    GtG = np.dot(G, G)   # Precompute normalisation
    
    amplitude_trace = np.zeros(T)
    
    # ---- Time loop (projection only) ----
    for t in range(T):
        patch = movie_corr[t, y_min:y_max, x_min:x_max].ravel()
        amplitude_trace[t] = np.dot(G, patch) / GtG
    
    # ---- ΔF/F normalisation (baseline 1–5 s) ----
    baseline_idx_dff = np.where(
        (time >= baseline_window_dff[0]) &
        (time <= baseline_window_dff[1])
    )[0]
    
    F0 = np.median(amplitude_trace[baseline_idx_dff])
    dff = (amplitude_trace - F0) / F0
    
    dff_traces.append(dff)

# ============================================================
# 12. PLOT TOP 5 TRACES
# ============================================================
plt.figure(figsize=(10,6))

for i,trace in enumerate(dff_traces):
    plt.plot(time, trace, label=f"Synapse {i+1}")

plt.xlabel("Time (s)")
plt.ylabel("ΔF/F (baseline 1–5 s)")
plt.title("Top 5 Synapse Responses")
plt.legend()
plt.tight_layout()
plt.show()

# ============================================================
# 13. COMPUTE ΔF/F FOR ALL SYNAPSES (Stable Projection)
# ============================================================

all_dff_traces = []

for y0,x0,sigma_fit,_ in synapses:
    
    if np.isnan(sigma_fit):
        sigma_fit = 1.5
    
    patch_radius = int(np.ceil(3*sigma_fit))
    
    y_min = max(0,y0-patch_radius)
    y_max = min(Y,y0+patch_radius+1)
    x_min = max(0,x0-patch_radius)
    x_max = min(X,x0+patch_radius+1)
    
    yy,xx = np.indices((y_max-y_min, x_max-x_min))
    
    G = np.exp(
        -((xx-(x0-x_min))**2 + (yy-(y0-y_min))**2) /
        (2*sigma_fit**2)
    )
    
    G = G.ravel()
    GtG = np.dot(G, G)
    
    amplitude_trace = np.zeros(T)
    
    for t in range(T):
        patch = movie_corr[t, y_min:y_max, x_min:x_max].ravel()
        amplitude_trace[t] = np.dot(G, patch) / GtG
    
    baseline_idx_dff = np.where(
        (time >= baseline_window_dff[0]) &
        (time <= baseline_window_dff[1])
    )[0]
    
    F0 = np.median(amplitude_trace[baseline_idx_dff])
    dff = (amplitude_trace - F0) / F0
    
    all_dff_traces.append(dff)

all_dff_traces = np.array(all_dff_traces)  # shape: (n_synapses, T)

# ============================================================
# 14. SORT SYNAPSES BY RESPONSE MAGNITUDE
# ============================================================

strength = np.max(np.abs(all_dff_traces), axis=1)
order = np.argsort(strength)[::-1]

sorted_traces = all_dff_traces[order]

# ============================================================
# 15. PLOT RESPONSE MATRIX (Each Row = Synapse)
# ============================================================

plt.figure(figsize=(10,6))

vmax = np.percentile(np.abs(sorted_traces), 99)
vmin = -vmax

plt.imshow(
    sorted_traces,
    aspect='auto',
    cmap='gray',
    vmin=vmin,
    vmax=vmax
)

# Convert x-axis to seconds
xticks = np.linspace(0, T-1, 6)
plt.xticks(xticks, np.round(xticks/fs,2))

plt.xlabel("Time (s)")
plt.ylabel("Synapses (sorted by |response|)")
plt.title("All Synapse ΔF/F Responses")
plt.colorbar(label="ΔF/F")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import tifffile as tiff
import matplotlib.pyplot as plt

from scipy.optimize import curve_fit
from scipy.ndimage import shift, gaussian_filter
from scipy.stats import ks_2samp
from skimage.registration import phase_cross_correlation
from skimage.feature import peak_local_max

# ============================================================
# PARAMETERS (YOUR WORKING ONES)
# ============================================================
fs = 20.0

baseline_windows = [(0,5), (55,60)]
transition_window_sec = 0.75
threshold_percentile = 70
min_distance = 2
roi_radius = 3

sigma_smooth = 0.8
edge_margin = 5
patch_radius = 4
alpha = 0.05

baseline_window_dff = (1,5)

# ============================================================
# 1. LOAD
# ============================================================
movie_raw = tiff.imread("movie1.tif")
stim_movie = tiff.imread("stim1.tif")

T, Y, X = movie_raw.shape
time = np.arange(T)/fs

# ============================================================
# 2. REGISTRATION
# ============================================================
reference = movie_raw.mean(axis=0)
movie_reg = np.zeros_like(movie_raw)

for i in range(T):
    shift_est,_,_ = phase_cross_correlation(
        reference, movie_raw[i], upsample_factor=10
    )
    movie_reg[i] = shift(movie_raw[i], shift_est)

# ============================================================
# 3. BLEACH CORRECTION
# ============================================================
frame_mean = movie_reg.mean(axis=(1,2))

def exp_decay(t,A,tau,C):
    return A*np.exp(-t/tau)+C

p0 = [frame_mean[0]-frame_mean[-1], T/fs/2, frame_mean[-1]]
params,_ = curve_fit(exp_decay,time,frame_mean,p0=p0,maxfev=10000)
fit_curve = exp_decay(time,*params)

movie_corr = movie_reg / np.maximum(fit_curve[:,None,None],1e-8)

# ============================================================
# 4. STIM TRANSITIONS
# ============================================================
stim_trace = stim_movie.mean(axis=(1,2))
stim_binary = stim_trace > np.median(stim_trace)
transitions = np.where(np.diff(stim_binary.astype(int))!=0)[0]

baseline_idx=[]
for s,e in baseline_windows:
    baseline_idx.extend(range(int(s*fs),int(e*fs)))
baseline_idx=np.array(baseline_idx)

transition_frames = int(transition_window_sec*fs)
transition_idx=[]
for idx in transitions:
    start=idx+1
    end=start+transition_frames
    if end<T:
        transition_idx.extend(range(start,end))
transition_idx=np.array(transition_idx)

# ============================================================
# 5. ΔF MAP
# ============================================================
mean_base = movie_corr[baseline_idx].mean(axis=0)
mean_trans = movie_corr[transition_idx].mean(axis=0)
delta_map = gaussian_filter(mean_trans - mean_base, sigma=sigma_smooth)

# ============================================================
# 6. PEAK DETECTION
# ============================================================
threshold_abs = np.percentile(delta_map, threshold_percentile)

coords = peak_local_max(
    delta_map,
    min_distance=min_distance,
    threshold_abs=threshold_abs,
    exclude_border=False
)

filtered_coords=[]
for y0,x0 in coords:
    if (y0>edge_margin and y0<Y-edge_margin and
        x0>edge_margin and x0<X-edge_margin):
        filtered_coords.append((y0,x0))

print("Candidate peaks:",len(filtered_coords))

# ============================================================
# 7. KS + FDR DETECTION
# ============================================================
pvals=[]
peaks=[]
rr_full,cc_full = np.indices((Y,X))

for y0,x0 in filtered_coords:
    mask = ((rr_full-y0)**2+(cc_full-x0)**2)<=roi_radius**2
    
    base_vals = movie_corr[baseline_idx][:,mask].mean(axis=1)
    trans_vals = movie_corr[transition_idx][:,mask].mean(axis=1)
    
    D,p = ks_2samp(base_vals,trans_vals)
    
    pvals.append(p)
    peaks.append((y0,x0,D,p))

pvals=np.array(pvals)

significant=[]
if len(pvals)>0:
    idx_sorted=np.argsort(pvals)
    sorted_p=pvals[idx_sorted]
    m=len(sorted_p)
    threshold_line=alpha*np.arange(1,m+1)/m
    significant_mask=sorted_p<=threshold_line
    
    if np.any(significant_mask):
        max_i=np.where(significant_mask)[0].max()
        p_cutoff=sorted_p[max_i]
        for i,(y0,x0,D,p) in enumerate(peaks):
            if p<=p_cutoff:
                significant.append((y0,x0))

print("Significant synapses:",len(significant))

# ============================================================
# 8. GAUSSIAN FIT (MORPHOLOGY ONLY)
# ============================================================
def gaussian2d(coords,A,x0,y0,sigma,B):
    x,y = coords
    return A*np.exp(-((x-x0)**2+(y-y0)**2)/(2*sigma**2))+B

synapses=[]

for y0,x0 in significant:
    
    sigma_fit = 1.5
    
    y_min=max(0,y0-patch_radius)
    y_max=min(Y,y0+patch_radius+1)
    x_min=max(0,x0-patch_radius)
    x_max=min(X,x0+patch_radius+1)
    
    patch = delta_map[y_min:y_max, x_min:x_max]
    yy,xx = np.indices(patch.shape)
    
    try:
        popt,_ = curve_fit(
            gaussian2d,
            (xx,yy),
            patch.ravel(),
            p0=[patch.max(), patch.shape[1]/2, patch.shape[0]/2, 1.5, patch.min()],
            maxfev=5000
        )
        sigma_fit = abs(popt[3])
    except:
        pass
    
    rr,cc = np.indices((Y,X))
    mask = ((rr-y0)**2+(cc-x0)**2)<=roi_radius**2
    F0 = movie_corr[baseline_idx][:,mask].mean()
    F1 = movie_corr[transition_idx][:,mask].mean()
    dff_detect = (F1-F0)/F0
    
    synapses.append((y0,x0,sigma_fit,dff_detect))

# ============================================================
# 9. PLOT SYNAPSE LOCATIONS (Highlight Top 5)
# ============================================================

# Extract original indices of top 5
top_indices = [syn_index for syn_index,_,_,_,_ in top_synapses]

plt.figure(figsize=(8,4))

vmin = np.percentile(delta_map,5)
vmax = np.percentile(delta_map,99)

plt.imshow(delta_map, cmap='gray', vmin=vmin, vmax=vmax)
plt.title(f"Detected Synapses (n={len(synapses)})")
plt.axis('off')

for i,(y0,x0,sigma_fit,dff) in enumerate(synapses,1):
    
    if i in top_indices:
        plt.scatter(x0, y0, s=120, c='cyan', edgecolors='black', linewidths=1.5)
    else:
        plt.scatter(x0, y0, s=60, c='red')
    
    plt.text(x0, y0, str(i),
             color='yellow',
             fontsize=9,
             ha='center',
             va='center',
             fontweight='bold')

plt.tight_layout()
plt.show()

# ============================================================
# 10. SELECT TOP 5 (Preserve Original Synapse Numbers)
# ============================================================

# Attach original index to each synapse
indexed_synapses = [
    (i+1, y0, x0, sigma_fit, dff_detect)
    for i,(y0,x0,sigma_fit,dff_detect) in enumerate(synapses)
]

# Sort by detection ΔF/F magnitude
indexed_synapses_sorted = sorted(
    indexed_synapses,
    key=lambda x: abs(x[4]),
    reverse=True
)

top_synapses = indexed_synapses_sorted[:5]

print("Top 5 synapses (original numbering):")
for syn_index,y,x,sigma,dff in top_synapses:
    print(f"Synapse {syn_index}: (y={y}, x={x}), sigma={sigma:.2f}")

# ============================================================
# 11. COMPUTE ΔF/F TRACES (Optimised Projection)
# ============================================================

dff_traces = []
labels = []

for syn_index,y0,x0,sigma_fit,_ in top_synapses:
    
    if np.isnan(sigma_fit):
        sigma_fit = 1.5
    
    patch_radius = int(np.ceil(3*sigma_fit))
    
    y_min = max(0,y0-patch_radius)
    y_max = min(Y,y0+patch_radius+1)
    x_min = max(0,x0-patch_radius)
    x_max = min(X,x0+patch_radius+1)
    
    yy,xx = np.indices((y_max-y_min, x_max-x_min))
    
    G = np.exp(
        -((xx-(x0-x_min))**2 + (yy-(y0-y_min))**2) /
        (2*sigma_fit**2)
    )
    
    G = G.ravel()
    GtG = np.dot(G, G)
    
    amplitude_trace = np.zeros(T)
    
    for t in range(T):
        patch = movie_corr[t, y_min:y_max, x_min:x_max].ravel()
        amplitude_trace[t] = np.dot(G, patch) / GtG
    
    baseline_idx_dff = np.where(
        (time >= baseline_window_dff[0]) &
        (time <= baseline_window_dff[1])
    )[0]
    
    F0 = np.median(amplitude_trace[baseline_idx_dff])
    dff = (amplitude_trace - F0) / F0
    
    dff_traces.append(dff)
    labels.append(f"Synapse {syn_index}")

# ============================================================
# 12. PLOT TOP 5 TRACES (Correct Labelling)
# ============================================================

plt.figure(figsize=(10,6))

for trace,label in zip(dff_traces, labels):
    plt.plot(time, trace, label=label)

plt.xlabel("Time (s)")
plt.ylabel("ΔF/F (baseline 1–5 s)")
plt.title("Top 5 Synapse Responses")
plt.legend()
plt.tight_layout()
plt.show()

# ============================================================
# 13. COMPUTE ΔF/F FOR ALL SYNAPSES (Stable Projection)
# ============================================================

all_dff_traces = []

for y0,x0,sigma_fit,_ in synapses:
    
    if np.isnan(sigma_fit):
        sigma_fit = 1.5
    
    patch_radius = int(np.ceil(3*sigma_fit))
    
    y_min = max(0,y0-patch_radius)
    y_max = min(Y,y0+patch_radius+1)
    x_min = max(0,x0-patch_radius)
    x_max = min(X,x0+patch_radius+1)
    
    yy,xx = np.indices((y_max-y_min, x_max-x_min))
    
    G = np.exp(
        -((xx-(x0-x_min))**2 + (yy-(y0-y_min))**2) /
        (2*sigma_fit**2)
    )
    
    G = G.ravel()
    GtG = np.dot(G, G)
    
    amplitude_trace = np.zeros(T)
    
    for t in range(T):
        patch = movie_corr[t, y_min:y_max, x_min:x_max].ravel()
        amplitude_trace[t] = np.dot(G, patch) / GtG
    
    baseline_idx_dff = np.where(
        (time >= baseline_window_dff[0]) &
        (time <= baseline_window_dff[1])
    )[0]
    
    F0 = np.median(amplitude_trace[baseline_idx_dff])
    dff = (amplitude_trace - F0) / F0
    
    all_dff_traces.append(dff)

all_dff_traces = np.array(all_dff_traces)  # shape: (n_synapses, T)

# ============================================================
# 14. SORT SYNAPSES BY RESPONSE MAGNITUDE
# ============================================================

strength = np.max(np.abs(all_dff_traces), axis=1)
order = np.argsort(strength)[::-1]

sorted_traces = all_dff_traces[order]

# ============================================================
# 15. PLOT RESPONSE MATRIX (Each Row = Synapse)
# ============================================================

plt.figure(figsize=(10,6))

vmax = np.percentile(np.abs(sorted_traces), 99)
vmin = -vmax

plt.imshow(
    sorted_traces,
    aspect='auto',
    cmap='gray',
    vmin=vmin,
    vmax=vmax
)

# Convert x-axis to seconds
xticks = np.linspace(0, T-1, 6)
plt.xticks(xticks, np.round(xticks/fs,2))

plt.xlabel("Time (s)")
plt.ylabel("Synapses (sorted by |response|)")
plt.title("All Synapse ΔF/F Responses")
plt.colorbar(label="ΔF/F")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import tifffile as tiff
import matplotlib.pyplot as plt

from scipy.optimize import curve_fit
from scipy.ndimage import shift, gaussian_filter
from scipy.stats import ks_2samp
from skimage.registration import phase_cross_correlation
from skimage.feature import peak_local_max

# ============================================================
# PARAMETERS (YOUR WORKING ONES)
# ============================================================
fs = 20.0

baseline_windows = [(0,5), (55,60)]
transition_window_sec = 1
threshold_percentile = 70
min_distance = 2
roi_radius = 2.5

sigma_smooth = 0.4
edge_margin = 3
patch_radius = 2
alpha = 0.05

baseline_window_dff = (1,5)

# ============================================================
# 1. LOAD
# ============================================================
movie_raw = tiff.imread("movie1.tif")
stim_movie = tiff.imread("stim1.tif")

T, Y, X = movie_raw.shape
time = np.arange(T)/fs

# ============================================================
# 2. REGISTRATION
# ============================================================
reference = movie_raw.mean(axis=0)
movie_reg = np.zeros_like(movie_raw)

for i in range(T):
    shift_est,_,_ = phase_cross_correlation(
        reference, movie_raw[i], upsample_factor=10
    )
    movie_reg[i] = shift(movie_raw[i], shift_est)

# ============================================================
# 3. BLEACH CORRECTION
# ============================================================
frame_mean = movie_reg.mean(axis=(1,2))

def exp_decay(t,A,tau,C):
    return A*np.exp(-t/tau)+C

p0 = [frame_mean[0]-frame_mean[-1], T/fs/2, frame_mean[-1]]
params,_ = curve_fit(exp_decay,time,frame_mean,p0=p0,maxfev=10000)
fit_curve = exp_decay(time,*params)

movie_corr = movie_reg / np.maximum(fit_curve[:,None,None],1e-8)

# ============================================================
# 4. STIM TRANSITIONS
# ============================================================
stim_trace = stim_movie.mean(axis=(1,2))
stim_binary = stim_trace > np.median(stim_trace)
transitions = np.where(np.diff(stim_binary.astype(int))!=0)[0]

baseline_idx=[]
for s,e in baseline_windows:
    baseline_idx.extend(range(int(s*fs),int(e*fs)))
baseline_idx=np.array(baseline_idx)

transition_frames = int(transition_window_sec*fs)
transition_idx=[]
for idx in transitions:
    start=idx+1
    end=start+transition_frames
    if end<T:
        transition_idx.extend(range(start,end))
transition_idx=np.array(transition_idx)

# ============================================================
# 5. ΔF MAP
# ============================================================
mean_base = movie_corr[baseline_idx].mean(axis=0)
mean_trans = movie_corr[transition_idx].mean(axis=0)
delta_map = gaussian_filter(mean_trans - mean_base, sigma=sigma_smooth)

# ============================================================
# 6. PEAK DETECTION
# ============================================================
threshold_abs = np.percentile(delta_map, threshold_percentile)

coords = peak_local_max(
    delta_map,
    min_distance=min_distance,
    threshold_abs=threshold_abs,
    exclude_border=False
)

filtered_coords=[]
for y0,x0 in coords:
    if (y0>edge_margin and y0<Y-edge_margin and
        x0>edge_margin and x0<X-edge_margin):
        filtered_coords.append((y0,x0))

print("Candidate peaks:",len(filtered_coords))

# ============================================================
# 7. KS + FDR DETECTION
# ============================================================
pvals=[]
peaks=[]
rr_full,cc_full = np.indices((Y,X))

for y0,x0 in filtered_coords:
    mask = ((rr_full-y0)**2+(cc_full-x0)**2)<=roi_radius**2
    
    base_vals = movie_corr[baseline_idx][:,mask].mean(axis=1)
    trans_vals = movie_corr[transition_idx][:,mask].mean(axis=1)
    
    D,p = ks_2samp(base_vals,trans_vals)
    
    pvals.append(p)
    peaks.append((y0,x0,D,p))

pvals=np.array(pvals)

significant=[]
if len(pvals)>0:
    idx_sorted=np.argsort(pvals)
    sorted_p=pvals[idx_sorted]
    m=len(sorted_p)
    threshold_line=alpha*np.arange(1,m+1)/m
    significant_mask=sorted_p<=threshold_line
    
    if np.any(significant_mask):
        max_i=np.where(significant_mask)[0].max()
        p_cutoff=sorted_p[max_i]
        for i,(y0,x0,D,p) in enumerate(peaks):
            if p<=p_cutoff:
                significant.append((y0,x0))

print("Significant synapses:",len(significant))

# ============================================================
# 8. GAUSSIAN FIT (MORPHOLOGY ONLY)
# ============================================================
def gaussian2d(coords,A,x0,y0,sigma,B):
    x,y = coords
    return A*np.exp(-((x-x0)**2+(y-y0)**2)/(2*sigma**2))+B

synapses=[]

for y0,x0 in significant:
    
    sigma_fit = 1.5
    
    y_min=max(0,y0-patch_radius)
    y_max=min(Y,y0+patch_radius+1)
    x_min=max(0,x0-patch_radius)
    x_max=min(X,x0+patch_radius+1)
    
    patch = delta_map[y_min:y_max, x_min:x_max]
    yy,xx = np.indices(patch.shape)
    
    try:
        popt,_ = curve_fit(
            gaussian2d,
            (xx,yy),
            patch.ravel(),
            p0=[patch.max(), patch.shape[1]/2, patch.shape[0]/2, 1.5, patch.min()],
            maxfev=5000
        )
        sigma_fit = abs(popt[3])
    except:
        pass
    
    rr,cc = np.indices((Y,X))
    mask = ((rr-y0)**2+(cc-x0)**2)<=roi_radius**2
    F0 = movie_corr[baseline_idx][:,mask].mean()
    F1 = movie_corr[transition_idx][:,mask].mean()
    dff_detect = (F1-F0)/F0
    
    synapses.append((y0,x0,sigma_fit,dff_detect))

# ============================================================
# 9. PLOT SYNAPSE LOCATIONS (Highlight Top 5)
# ============================================================

# Extract original indices of top 5
top_indices = [syn_index for syn_index,_,_,_,_ in top_synapses]

plt.figure(figsize=(8,4))

vmin = np.percentile(delta_map,5)
vmax = np.percentile(delta_map,99)

plt.imshow(delta_map, cmap='gray', vmin=vmin, vmax=vmax)
plt.title(f"Detected Synapses (n={len(synapses)})")
plt.axis('off')

for i,(y0,x0,sigma_fit,dff) in enumerate(synapses,1):
    
    if i in top_indices:
        plt.scatter(x0, y0, s=120, c='cyan', edgecolors='black', linewidths=1.5)
    else:
        plt.scatter(x0, y0, s=60, c='red')
    
    plt.text(x0, y0, str(i),
             color='yellow',
             fontsize=9,
             ha='center',
             va='center',
             fontweight='bold')

plt.tight_layout()
plt.show()

# ============================================================
# 10. SELECT TOP 5 (Preserve Original Synapse Numbers)
# ============================================================

# Attach original index to each synapse
indexed_synapses = [
    (i+1, y0, x0, sigma_fit, dff_detect)
    for i,(y0,x0,sigma_fit,dff_detect) in enumerate(synapses)
]

# Sort by detection ΔF/F magnitude
indexed_synapses_sorted = sorted(
    indexed_synapses,
    key=lambda x: abs(x[4]),
    reverse=True
)

top_synapses = indexed_synapses_sorted[:5]

print("Top 5 synapses (original numbering):")
for syn_index,y,x,sigma,dff in top_synapses:
    print(f"Synapse {syn_index}: (y={y}, x={x}), sigma={sigma:.2f}")

# ---- Build global spatial demixing matrix once ----

ys = [y for y,_,_,_ in synapses]
xs = [x for _,x,_,_ in synapses]

margin = 6
y_min = max(0, min(ys) - margin)
y_max = min(Y, max(ys) + margin)
x_min = max(0, min(xs) - margin)
x_max = min(X, max(xs) + margin)

subY = y_max - y_min
subX = x_max - x_min

# Build spatial template matrix
G_list = []

for y0,x0,sigma_fit,_ in synapses:
    
    if np.isnan(sigma_fit):
        sigma_fit = 1.5
    
    yy,xx = np.indices((subY,subX))
    
    G_i = np.exp(
        -((xx-(x0-x_min))**2 + (yy-(y0-y_min))**2) /
        (2*sigma_fit**2)
    )
    
    G_list.append(G_i.ravel())

G = np.column_stack(G_list)

#GtG = G.T @ G
#W = np.linalg.pinv(GtG) @ G.T

lambda_reg = 0.05   # small stabilisation

GtG = G.T @ G
GtG_reg = GtG + lambda_reg * np.eye(GtG.shape[0])

W = np.linalg.solve(GtG_reg, G.T)

#---- Extract amplitudes for all synapses simultaneously ----

A_all = np.zeros((len(synapses), T))

for t in range(T):
    frame = movie_corr[t, y_min:y_max, x_min:x_max].ravel()
    A_all[:,t] = W @ frame

# ---- Now extract only top-5 traces ----

dff_traces = []
labels = []

baseline_idx_dff = np.where(
    (time >= baseline_window_dff[0]) &
    (time <= baseline_window_dff[1])
)[0]

for syn_index,y0,x0,sigma_fit,_ in top_synapses:
    
    amp = A_all[syn_index-1]   # use original synapse index
    
    F0 = np.median(amp[baseline_idx_dff])
    dff = (amp - F0) / F0
    
    dff_traces.append(dff)
    labels.append(f"Synapse {syn_index}")

# ============================================================
# 12. PLOT TOP 5 TRACES (Correct Labelling)
# ============================================================

plt.figure(figsize=(10,6))

for trace,label in zip(dff_traces, labels):
    plt.plot(time, trace, label=label)

plt.xlabel("Time (s)")
plt.ylabel("ΔF/F (baseline 1–5 s)")
plt.title("Top 5 Synapse Responses")
plt.legend()
plt.tight_layout()
plt.show()

all_dff_traces = []

for i in range(len(synapses)):
    
    amp = A_all[i]
    
    F0 = np.median(amp[baseline_idx_dff])
    dff = (amp - F0) / F0
    
    all_dff_traces.append(dff)

all_dff_traces = np.array(all_dff_traces)

# ============================================================
# 14. SORT SYNAPSES BY RESPONSE MAGNITUDE
# ============================================================

strength = np.max(np.abs(all_dff_traces), axis=1)
order = np.argsort(strength)[::-1]

sorted_traces = all_dff_traces[order]

# ============================================================
# 15. PLOT RESPONSE MATRIX (Each Row = Synapse)
# ============================================================

plt.figure(figsize=(10,6))

vmax = np.percentile(np.abs(sorted_traces), 99)
vmin = -vmax

plt.imshow(
    sorted_traces,
    aspect='auto',
    cmap='gray',
    vmin=vmin,
    vmax=vmax
)

# Convert x-axis to seconds
xticks = np.linspace(0, T-1, 6)
plt.xticks(xticks, np.round(xticks/fs,2))

plt.xlabel("Time (s)")
plt.ylabel("Synapses (sorted by |response|)")
plt.title("All Synapse ΔF/F Responses")
plt.colorbar(label="ΔF/F")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import tifffile as tiff
import matplotlib.pyplot as plt

from scipy.optimize import curve_fit
from scipy.ndimage import shift, gaussian_filter
from scipy.stats import ks_2samp
from skimage.registration import phase_cross_correlation
from skimage.feature import peak_local_max

# ============================================================
# PARAMETERS (YOUR WORKING ONES)
# ============================================================
fs = 20.0

baseline_windows = [(0,5), (55,60)]
transition_window_sec = 1
threshold_percentile = 70
min_distance = 2
roi_radius = 4

sigma_smooth = 0.4
edge_margin = 3
patch_radius = 2
alpha = 0.05

baseline_window_dff = (1,5)


# ============================================================
# 1. LOAD (INTERLEAVED TIFF) + DEINTERLEAVE
# ============================================================

interleaved = tiff.imread("movie.tif")

T_total, Y, X = interleaved.shape

if T_total % 2 != 0:
    raise ValueError("Interleaved movie must have even number of frames.")

# Assume:
#   even indices  -> movie
#   odd indices   -> stim
# If reversed, swap slicing below.

movie_raw = interleaved[0::2]
stim_movie = interleaved[1::2]

T = movie_raw.shape[0]
time = np.arange(T) / fs

print("Total frames:", T_total)
print("Movie frames:", movie_raw.shape[0])
print("Stim frames:", stim_movie.shape[0])

T, Y, X = movie_raw.shape
time = np.arange(T)/fs

# ============================================================
# 2. REGISTRATION
# ============================================================
reference = movie_raw.mean(axis=0)
movie_reg = np.zeros_like(movie_raw)

for i in range(T):
    shift_est,_,_ = phase_cross_correlation(
        reference, movie_raw[i], upsample_factor=10
    )
    movie_reg[i] = shift(movie_raw[i], shift_est)

# ============================================================
# 3. BLEACH CORRECTION
# ============================================================
frame_mean = movie_reg.mean(axis=(1,2))

def exp_decay(t,A,tau,C):
    return A*np.exp(-t/tau)+C

p0 = [frame_mean[0]-frame_mean[-1], T/fs/2, frame_mean[-1]]
params,_ = curve_fit(exp_decay,time,frame_mean,p0=p0,maxfev=10000)
fit_curve = exp_decay(time,*params)

movie_corr = movie_reg / np.maximum(fit_curve[:,None,None],1e-8)

# ============================================================
# 4. STIM TRANSITIONS
# ============================================================
stim_trace = stim_movie.mean(axis=(1,2))
stim_binary = stim_trace > np.median(stim_trace)
transitions = np.where(np.diff(stim_binary.astype(int))!=0)[0]

baseline_idx=[]
for s,e in baseline_windows:
    baseline_idx.extend(range(int(s*fs),int(e*fs)))
baseline_idx=np.array(baseline_idx)

transition_frames = int(transition_window_sec*fs)
transition_idx=[]
for idx in transitions:
    start=idx+1
    end=start+transition_frames
    if end<T:
        transition_idx.extend(range(start,end))
transition_idx=np.array(transition_idx)

# ============================================================
# 5. ΔF MAP
# ============================================================
mean_base = movie_corr[baseline_idx].mean(axis=0)
mean_trans = movie_corr[transition_idx].mean(axis=0)
delta_map = gaussian_filter(mean_trans - mean_base, sigma=sigma_smooth)

# ============================================================
# 6. PEAK DETECTION
# ============================================================
threshold_abs = np.percentile(delta_map, threshold_percentile)

coords = peak_local_max(
    delta_map,
    min_distance=min_distance,
    threshold_abs=threshold_abs,
    exclude_border=False
)

filtered_coords=[]
for y0,x0 in coords:
    if (y0>edge_margin and y0<Y-edge_margin and
        x0>edge_margin and x0<X-edge_margin):
        filtered_coords.append((y0,x0))

print("Candidate peaks:",len(filtered_coords))

# ============================================================
# 7. KS + FDR DETECTION
# ============================================================
pvals=[]
peaks=[]
rr_full,cc_full = np.indices((Y,X))

for y0,x0 in filtered_coords:
    mask = ((rr_full-y0)**2+(cc_full-x0)**2)<=roi_radius**2
    
    base_vals = movie_corr[baseline_idx][:,mask].mean(axis=1)
    trans_vals = movie_corr[transition_idx][:,mask].mean(axis=1)
    
    D,p = ks_2samp(base_vals,trans_vals)
    
    pvals.append(p)
    peaks.append((y0,x0,D,p))

pvals=np.array(pvals)

significant=[]
if len(pvals)>0:
    idx_sorted=np.argsort(pvals)
    sorted_p=pvals[idx_sorted]
    m=len(sorted_p)
    threshold_line=alpha*np.arange(1,m+1)/m
    significant_mask=sorted_p<=threshold_line
    
    if np.any(significant_mask):
        max_i=np.where(significant_mask)[0].max()
        p_cutoff=sorted_p[max_i]
        for i,(y0,x0,D,p) in enumerate(peaks):
            if p<=p_cutoff:
                significant.append((y0,x0))

print("Significant synapses:",len(significant))

# ============================================================
# 8. GAUSSIAN FIT (MORPHOLOGY ONLY)
# ============================================================
def gaussian2d(coords,A,x0,y0,sigma,B):
    x,y = coords
    return A*np.exp(-((x-x0)**2+(y-y0)**2)/(2*sigma**2))+B

synapses=[]

for y0,x0 in significant:
    
    sigma_fit = 1.5
    
    y_min=max(0,y0-patch_radius)
    y_max=min(Y,y0+patch_radius+1)
    x_min=max(0,x0-patch_radius)
    x_max=min(X,x0+patch_radius+1)
    
    patch = delta_map[y_min:y_max, x_min:x_max]
    yy,xx = np.indices(patch.shape)
    
    try:
        popt,_ = curve_fit(
            gaussian2d,
            (xx,yy),
            patch.ravel(),
            p0=[patch.max(), patch.shape[1]/2, patch.shape[0]/2, 1.5, patch.min()],
            maxfev=5000
        )
        sigma_fit = abs(popt[3])
    except:
        pass
    
    rr,cc = np.indices((Y,X))
    mask = ((rr-y0)**2+(cc-x0)**2)<=roi_radius**2
    F0 = movie_corr[baseline_idx][:,mask].mean()
    F1 = movie_corr[transition_idx][:,mask].mean()
    dff_detect = (F1-F0)/F0
    
    synapses.append((y0,x0,sigma_fit,dff_detect))

# ============================================================
# 9. SELECT TOP 5
# ============================================================

indexed_synapses = [
    (i+1, y0, x0, sigma_fit, dff_detect)
    for i,(y0,x0,sigma_fit,dff_detect) in enumerate(synapses)
]

indexed_synapses_sorted = sorted(
    indexed_synapses,
    key=lambda x: abs(x[4]),
    reverse=True
)

top_synapses = indexed_synapses_sorted[:5]

print("Top 5 synapses (original numbering):")
for syn_index,y,x,sigma,dff in top_synapses:
    print(f"Synapse {syn_index}: (y={y}, x={x}), sigma={sigma:.2f}")

# ============================================================
# 10. PLOT SYNAPSE LOCATIONS (NOW CORRECTLY PLACED)
# ============================================================

top_indices = [syn_index for syn_index,_,_,_,_ in top_synapses]

plt.figure(figsize=(8,4))

vmin = np.percentile(delta_map,5)
vmax = np.percentile(delta_map,99)

plt.imshow(delta_map, cmap='gray', vmin=vmin, vmax=vmax)
plt.title(f"Detected Synapses (n={len(synapses)})")
plt.axis('off')

for i,(y0,x0,sigma_fit,dff) in enumerate(synapses,1):
    if i in top_indices:
        plt.scatter(x0, y0, s=120, c='cyan', edgecolors='black', linewidths=1.5)
    else:
        plt.scatter(x0, y0, s=60, c='red')
    
    plt.text(x0, y0, str(i),
             color='yellow',
             fontsize=9,
             ha='center',
             va='center',
             fontweight='bold')

plt.tight_layout()
plt.show()

# ============================================================
# 11. RIDGE DEMIXING (UNCHANGED)
# ============================================================

ys = [y for y,_,_,_ in synapses]
xs = [x for _,x,_,_ in synapses]

margin = 6
y_min = max(0, min(ys) - margin)
y_max = min(Y, max(ys) + margin)
x_min = max(0, min(xs) - margin)
x_max = min(X, max(xs) + margin)

subY = y_max - y_min
subX = x_max - x_min

G_list = []

for y0,x0,sigma_fit,_ in synapses:
    if np.isnan(sigma_fit):
        sigma_fit = 1.5
    
    yy,xx = np.indices((subY,subX))
    G_i = np.exp(
        -((xx-(x0-x_min))**2 + (yy-(y0-y_min))**2) /
        (2*sigma_fit**2)
    )
    G_list.append(G_i.ravel())

G = np.column_stack(G_list)

lambda_reg = 0.05

GtG = G.T @ G
GtG_reg = GtG + lambda_reg * np.eye(GtG.shape[0])
W = np.linalg.solve(GtG_reg, G.T)

A_all = np.zeros((len(synapses), T))

for t in range(T):
    frame = movie_corr[t, y_min:y_max, x_min:x_max].ravel()
    A_all[:,t] = W @ frame


# ============================================================
# 1. Compute independent (non-demixed) traces
# ============================================================

independent_traces = []

baseline_idx_dff = np.where(
    (time >= baseline_window_dff[0]) &
    (time <= baseline_window_dff[1])
)[0]

for y0,x0,sigma_fit,_ in synapses:
    
    if np.isnan(sigma_fit):
        sigma_fit = 1.5
    
    rr,cc = np.indices((Y,X))
    mask = ((rr-y0)**2+(cc-x0)**2)<=roi_radius**2
    
    amp = movie_corr[:,mask].mean(axis=1)
    F0 = np.median(amp[baseline_idx_dff])
    dff = (amp - F0) / F0
    
    independent_traces.append(dff)

independent_traces = np.array(independent_traces)

# ============================================================
# 2. Ridge-demixed traces (already computed)
# ============================================================

ridge_traces = []

for i in range(len(synapses)):
    amp = A_all[i]
    F0 = np.median(amp[baseline_idx_dff])
    dff = (amp - F0) / F0
    ridge_traces.append(dff)

ridge_traces = np.array(ridge_traces)

# ============================================================
# 3. Compute pairwise correlations
# ============================================================

corr_ind = np.corrcoef(independent_traces)
corr_ridge = np.corrcoef(ridge_traces)

# ============================================================
# 4. Compute inter-synapse distances
# ============================================================

coords = np.array([(y,x) for y,x,_,_ in synapses])

dist_matrix = np.sqrt(
    (coords[:,None,0] - coords[None,:,0])**2 +
    (coords[:,None,1] - coords[None,:,1])**2
)

# ============================================================
# 5. Extract only neighbour pairs (< 6 pixels)
# ============================================================

distance_threshold = 6

mask_pairs = (dist_matrix > 0) & (dist_matrix < distance_threshold)

neighbor_corr_ind = corr_ind[mask_pairs]
neighbor_corr_ridge = corr_ridge[mask_pairs]

print("Mean neighbour correlation (Independent):",
      np.mean(neighbor_corr_ind))

print("Mean neighbour correlation (Ridge):",
      np.mean(neighbor_corr_ridge))


import matplotlib.gridspec as gridspec

# ============================================================
# TOP 5 TRACES + STIM TRACE
# ============================================================

fig = plt.figure(figsize=(10,7))
gs = gridspec.GridSpec(20, 1)

# ---- Main traces (95%)
ax_main = fig.add_subplot(gs[:19, 0])

for trace,label in zip(dff_traces, labels):
    ax_main.plot(time, trace, label=label)

ax_main.set_ylabel("ΔF/F (baseline 1–5 s)")
ax_main.set_title("Top 5 Synapse Responses")
ax_main.legend()

# ---- Stim trace (5%)
ax_stim = fig.add_subplot(gs[19, 0], sharex=ax_main)

stim_trace = stim_movie.mean(axis=(1,2))
stim_trace = (stim_trace - stim_trace.min()) / (stim_trace.max() - stim_trace.min())

ax_stim.plot(time, stim_trace, color='black', linewidth=1)
ax_stim.set_ylim(0,1)
ax_stim.axis('off')

plt.tight_layout()
plt.show()

# ============================================================
# 14. SORT SYNAPSES BY RESPONSE MAGNITUDE
# ============================================================

strength = np.max(np.abs(all_dff_traces), axis=1)
order = np.argsort(strength)[::-1]

sorted_traces = all_dff_traces[order]

# ============================================================
# RESPONSE MATRIX + STIM TRACE
# ============================================================

fig = plt.figure(figsize=(10,7))
gs = gridspec.GridSpec(20, 1)

# ---- Heatmap (95%)
ax_heat = fig.add_subplot(gs[:19, 0])

vmax = np.percentile(np.abs(sorted_traces), 99)
vmin = -vmax

im = ax_heat.imshow(
    sorted_traces,
    aspect='auto',
    cmap='gray',
    vmin=vmin,
    vmax=vmax
)

xticks = np.linspace(0, T-1, 6)
ax_heat.set_xticks(xticks)
ax_heat.set_xticklabels(np.round(xticks/fs,2))
ax_heat.set_ylabel("Synapses (sorted by |response|)")
ax_heat.set_title("All Synapse ΔF/F Responses")

# ---- Stim trace (5%)
ax_stim = fig.add_subplot(gs[19, 0], sharex=ax_heat)

stim_trace = stim_movie.mean(axis=(1,2))
stim_trace = (stim_trace - stim_trace.min()) / (stim_trace.max() - stim_trace.min())

ax_stim.plot(time, stim_trace, color='black', linewidth=1)
ax_stim.set_ylim(0,1)
ax_stim.axis('off')

plt.tight_layout()
plt.show()

plt.figure(figsize=(8,4))

plt.hist(neighbor_corr_ind, bins=20, alpha=0.5, label="Independent")
plt.hist(neighbor_corr_ridge, bins=20, alpha=0.5, label="Ridge")

plt.xlabel("Pairwise Correlation (Neighbouring Synapses)")
plt.ylabel("Count")
plt.legend()
plt.title("Spatial Cross-Talk Comparison")
plt.tight_layout()
plt.show()
```
